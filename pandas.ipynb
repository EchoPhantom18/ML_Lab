19.	Conditional Column
o	Cre	F: < 40
ate a Grade column:
	A: ≥ 90
	B: 75–89
	C: 60–74

```python
student["Grade"] = ""
student.loc[student["Marks"]>=90, "Grade"]="A"
student.loc[(student["Marks"]>=75) & (student["Marks"]<=89) , "Grade"]= "B" 
student.loc[(student["Marks"]>=60) & (student["Marks"]<=74), "Grade"]="C"
student.loc[(student["Marks"]>=40)&(student["Marks"]<59), "Grade"]="D"
student.loc[(student["Marks"]<40), "Grade"]="F"
```

28.	Pivot Table
o	Given sales data containing Employee, Product, Region, and Sales, create a pivot table showing total sales by region and product.


```python
import pandas as pd

sales = pd.DataFrame({
    "Employee": ["Anmol", "Rahul", "Aman", "Priya", "Riya",
                 "Karan", "Neha", "Arjun", "Simran", "Vivek"],
    "Product": ["Laptop", "Phone", "Laptop", "Tablet", "Phone",
                "Laptop", "Tablet", "Phone", "Laptop", "Tablet"],
    "Region": ["North", "South", "North", "East", "South",
               "West", "East", "West", "North", "South"],
    "Sales": [50000, 30000, 45000, 25000, 35000,
              60000, 28000, 32000, 55000, 22000]
})

print(sales)
```

```python
pivot=pd.pivot_table(
    sales,
    values="Sales",
    index="Region",
    columns="Product",
    aggfunc="sum",
    fill_value=0  #replces nan with 0
)
print(pivot)
```
