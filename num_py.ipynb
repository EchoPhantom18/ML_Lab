```python
'''Find prime numbers in an array'''
import numpy as np

a = np.array([2, 4, 5, 7, 9, 11, 15, 17])

prime = []

for n in a:
    if n > 1:
        is_prime = True

        for i in range(2, int(np.sqrt(n)) + 1):
            if n % i == 0:
                is_prime = False
                break

        if is_prime:
            prime.append(n)

print("Prime numbers:", prime)
```

```python
'''Random marks — 100 students, 5 subjects'''
import numpy as np

marks = np.random.randint(0, 101, (100, 5))

# Student-wise average
student_avg = np.mean(marks, axis=1)

# Subject-wise average
subject_avg = np.mean(marks, axis=0)

# Highest scorer
highest = np.argmax(student_avg)

# Lowest scorer
lowest = np.argmin(student_avg)

print("Student-wise average:")
print(student_avg)

print("Subject-wise average:")
print(subject_avg)

print("Highest scorer: Student", highest + 1)
print("Lowest scorer: Student", lowest + 1)
```

```python
'''Sales data — 12 months, 5 products'''
import numpy as np

sales = np.random.randint(100, 1000, (12, 5))

# Total sales of each product
product_sales = np.sum(sales, axis=0)

# Best-selling product
best_product = np.argmax(product_sales)

# Total sales of each month
monthly_sales = np.sum(sales, axis=1)

# Highest sales month
best_month = np.argmax(monthly_sales)

# Average monthly sales
average_monthly_sales = np.mean(monthly_sales)

print("Best-selling product: Product", best_product + 1)
print("Highest sales month: Month", best_month + 1)
print("Average monthly sales:", average_monthly_sales)
```

```python
'''Employee salary analysis'''
import numpy as np

salary = np.random.randint(20000, 100000, 20)

average = np.mean(salary)

# Employees earning above average
above_average = salary[salary > average]

# Highest salary
highest = np.max(salary)

# Lowest salary
lowest = np.min(salary)

# Top 10 salaries
top_10 = np.sort(salary)[-10:][::-1]

print("Salaries:")
print(salary)

print("Average salary:", average)

print("Employees earning above average:")
print(above_average)

print("Highest salary:", highest)

print("Lowest salary:", lowest)

print("Top 10 salaries:")
print(top_10)
```
